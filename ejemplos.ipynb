## Tipos de datos

Algunos de los tipos de datos que se utilizan en Python son los siguientes:

- **Enteros**: números sin decimales, como 1, 2, 3, -5, etc.
- **Flotantes**: números con decimales, como 1.5, 2.0, -3.14, etc.
- **Cadenas de texto**: secuencias de caracteres, como "Hola", "Mundo", "123", etc.
- **Booleanos**: valores de verdad, que pueden ser `True` o `False`.
- **Listas**: colecciones ordenadas de elementos, que pueden ser de diferentes tipos, como `[1, 2, 3]`, `["a", "b", "c"]`, etc.
- **Diccionarios**: colecciones de pares clave-valor, como `{"nombre": "Juan", "edad": 30}`.
- **Tuplas**: colecciones ordenadas e inmutables de elementos, como `(1, 2, 3)`, `("a", "b", "c")`, etc.
- **Conjuntos**: colecciones desordenadas de elementos únicos, como `{1, 2, 3}`, `{"a", "b", "c"}`, etc.

Comenzaremos con los números, porque finalmente todo se reduce a eso, y es lo que más se utiliza en la programación y análisis estadístico. En Python, los números se pueden representar de varias formas, pero los más comunes son los enteros y los flotantes.

En la siguiente celda de código podemos ver algunos ejemplos de cómo se pueden utilizar los números en Python:

```python
# Ejemplos de números en Python
# Enteros
entero_positivo = 10
entero_negativo = -5
# Flotantes
flotante_positivo = 10.5
flotante_negativo = -3.14
```

## Operadores aritméticos

Son los usuales de suma, resta, multiplicación, división, módulo, potencia y división entera. A continuación se muestran algunos ejemplos de cómo se pueden utilizar estos operadores en Python:

```python
# Ejemplos de operadores aritméticos en Python
# Suma
suma = 5 + 3  # Resultado: 8
# Resta
resta = 5 - 3  # Resultado: 2
# Multiplicación
multiplicacion = 5 * 3  # Resultado: 15
# División
division = 5 / 3  # Resultado: 1.666666666666666
# Módulo
modulo = 5 % 3  # Resultado: 2
# Potencia
potencia = 5 ** 3  # Resultado: 125
# División entera
division_entera = 5 // 3  # Resultado: 1
```

### Operaciones con enteros y flotantes

Las operaciones son las mismas, pero hay que tener en cuenta que si se realiza una operación entre un entero y un flotante, el resultado será un flotante, o sea, si se realiza una suma entre un entero y un flotante, el resultado será un flotante.

En el caso de la división, si se realiza una división entre dos enteros, el resultado será un flotante.

En el caso delmódulo, el resultado siempreserá un número entero, ya que el módulo es el resto de la división entre dos números enteros.

## Operaciones combinadas

Se matiene el orden de precedencia, es decir, primero se realizan las operaciones dentro de los paréntesis, luego las multiplicaciones y divisiones, y finalmente las sumas y restas. En caso de que haya operaciones del mismo nivel de precedencia, se realizan de izquierda a derecha, lo que en muchas partes de conoce como PEMDAS (Paréntesis, Exponentes, Multiplicación y División, Adición y Sustracción) o PAPOMUDAS (Paréntesis, Potencias, Multiplicación y División, Adición y Sustracción).

### Ejemplos de operaciones combinadas

````python
-5 + 7**3 / (3 + 2.5)
````

````python
# En el caso de la raíz, se utiliza la función pow(base,exponente) y el exponente se reemplaza por el índice de la raíz, como fracción o decimal
pow(2,1/2)
````

````python
pow((-3 + 8)/(3 - 5) + 10,1/2)
````
